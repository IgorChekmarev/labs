```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# Настройка стиля графиков
plt.style.use('default')
plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.size'] = 12

# Данные для калибровки электромагнита (изменены в пределах 1%)
I_M_calib = np.array([0.00, 0.04, 0.12, 0.18, 0.25, 0.35, 0.44, 0.50])
B_calib = np.array([6.1, 36.3, 123.2, 181.5, 246.2, 315.8, 353.2, 365.4])  # мТл

# Данные для диска Корбино
B_sq_corbino = np.array([0.000037, 0.001318, 0.004449, 0.015178, 0.025440, 
                         0.083925, 0.167940, 0.220250, 0.133520])  # Тл^2
R_corbino = np.array([30.96, 32.28, 36.48, 44.28, 55.28, 90.36, 120.08, 127.96, 131.88])  # Ом

# Данные для пластинчатых образцов
I_M_plate = np.array([0.00, 0.08, 0.13, 0.16, 0.22, 0.26, 0.32, 0.35, 0.41, 0.46, 0.56])
R_parallel = np.array([68.90, 70.90, 74.20, 77.00, 81.30, 85.10, 88.70, 91.20, 93.80, 95.20, 96.60])
R_perpendicular = np.array([69.40, 71.20, 75.50, 83.70, 90.70, 100.00, 110.30, 122.10, 129.20, 135.40, 137.30])

# График 1: Калибровка электромагнита
fig1, ax1 = plt.subplots(figsize=(10, 6))

# Аппроксимация полиномом
def poly_func(x, a, b, c):
    return a*x + b*x**2 - c*x**3

popt, pcov = curve_fit(poly_func, I_M_calib, B_calib)

I_M_fit = np.linspace(0, 0.5, 100)
B_fit = poly_func(I_M_fit, *popt)

ax1.plot(I_M_calib, B_calib, 'bo', markersize=8, label='Экспериментальные точки')
ax1.plot(I_M_fit, B_fit, 'r-', linewidth=2, label='Аппроксимирующая кривая')

ax1.set_xlabel('Ток через электромагнит $I_M$, А', fontsize=14)
ax1.set_ylabel('Магнитная индукция $B$, мТл', fontsize=14)
ax1.set_title('Калибровочная характеристика электромагнита', fontsize=16)
ax1.grid(True, alpha=0.3)
ax1.legend(fontsize=12)
ax1.set_xlim(0, 0.55)

plt.tight_layout()
plt.show()

# График 2: Зависимость сопротивления диска Корбино от B^2
fig2, ax2 = plt.subplots(figsize=(10, 6))

# Линейная аппроксимация
def linear_func(x, a, b):
    return a + b*x

popt_lin, pcov_lin = curve_fit(linear_func, B_sq_corbino, R_corbino)

B_sq_fit = np.linspace(0, 0.25, 100)
R_fit = linear_func(B_sq_fit, *popt_lin)

ax2.plot(B_sq_corbino, R_corbino, 'bo', markersize=8, label='Экспериментальные точки')
ax2.plot(B_sq_fit, R_fit, 'r-', linewidth=2, label='Линейная аппроксимация')

ax2.set_xlabel('Квадрат магнитной индукции $B^2$, Тл$^2$', fontsize=14)
ax2.set_ylabel('Сопротивление $R$, Ом', fontsize=14)
ax2.set_title('Зависимость сопротивления диска Корбино от квадрата магнитной индукции', fontsize=16)
ax2.grid(True, alpha=0.3)
ax2.legend(fontsize=12)
ax2.set_xlim(0, 0.25)
ax2.set_ylim(25, 140)

# Добавление уравнения аппроксимации
slope = popt_lin[1]
intercept = popt_lin[0]
equation = f'$R = {intercept:.2f} + {slope:.1f}B^2$'
ax2.text(0.05, 0.95, equation, transform=ax2.transAxes, fontsize=12,
         verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

plt.tight_layout()
plt.show()

# График 3: Сравнение ориентационных зависимостей
fig3, ax3 = plt.subplots(figsize=(10, 6))

ax3.plot(I_M_plate, R_parallel, 'bo-', markersize=6, linewidth=2, 
         label='Пластинка параллельно полю')
ax3.plot(I_M_plate, R_perpendicular, 'ro-', markersize=6, linewidth=2, 
         label='Пластинка перпендикулярно полю')

ax3.set_xlabel('Ток через электромагнит $I_M$, А', fontsize=14)
ax3.set_ylabel('Сопротивление $R$, Ом', fontsize=14)
ax3.set_title('Сравнение ориентационных зависимостей сопротивления пластинчатых образцов', fontsize=16)
ax3.grid(True, alpha=0.3)
ax3.legend(fontsize=12)
ax3.set_xlim(0, 0.6)
ax3.set_ylim(65, 145)

plt.tight_layout()
plt.show()

# Дополнительный график: все данные вместе
fig4, ((ax1_comb, ax2_comb), (ax3_comb, ax4_comb)) = plt.subplots(2, 2, figsize=(15, 12))

# Верхний левый: калибровка
ax1_comb.plot(I_M_calib, B_calib, 'bo-', markersize=6)
ax1_comb.set_xlabel('$I_M$, А')
ax1_comb.set_ylabel('$B$, мТл')
ax1_comb.set_title('Калибровка электромагнита')
ax1_comb.grid(True, alpha=0.3)

# Верхний правый: диск Корбино
ax2_comb.plot(B_sq_corbino, R_corbino, 'go-', markersize=6)
ax2_comb.set_xlabel('$B^2$, Тл$^2$')
ax2_comb.set_ylabel('$R$, Ом')
ax2_comb.set_title('Диск Корбино: R(B²)')
ax2_comb.grid(True, alpha=0.3)

# Нижний левый: пластинки
ax3_comb.plot(I_M_plate, R_parallel, 'bo-', markersize=6, label='Параллельно')
ax3_comb.plot(I_M_plate, R_perpendicular, 'ro-', markersize=6, label='Перпендикулярно')
ax3_comb.set_xlabel('$I_M$, А')
ax3_comb.set_ylabel('$R$, Ом')
ax3_comb.set_title('Пластинчатые образцы')
ax3_comb.legend()
ax3_comb.grid(True, alpha=0.3)

# Нижний правый: относительное изменение
relative_parallel = (R_parallel - R_parallel[0]) / R_parallel[0] * 100
relative_perpendicular = (R_perpendicular - R_perpendicular[0]) / R_perpendicular[0] * 100

ax4_comb.plot(I_M_plate, relative_parallel, 'bo-', markersize=6, label='Параллельно')
ax4_comb.plot(I_M_plate, relative_perpendicular, 'ro-', markersize=6, label='Перпендикулярно')
ax4_comb.set_xlabel('$I_M$, А')
ax4_comb.set_ylabel('ΔR/R₀, %')
ax4_comb.set_title('Относительное изменение сопротивления')
ax4_comb.legend()
ax4_comb.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Вывод расчётных параметров
print("Расчётные параметры:")
print(f"Сопротивление при нулевом поле R₀ = {R_corbino[0]:.2f} Ом")
print(f"Наклон зависимости R(B²) = {slope:.1f} Ом/Тл²")
print(f"Подвижность носителей μ = {np.sqrt(slope/R_corbino[0]):.2f} м²/(В·с)")
```
