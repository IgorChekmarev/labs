```python
import matplotlib.pyplot as plt
import numpy as np

# Данные
x1 = np.array([0, 0.07, 0.16, 0.21, 0.28, 0.36,0.45, 0.51])  # I, А
y = np.array([8.9, 74.7, 161.9, 213, 284.3, 327, 356.8, 368.9])*0.001  # B, Тл
t = np.full_like(y, 0.5)  # Погрешность (массив той же длины, что y)

# Линия для интерполяции
x_line = np.linspace(min(x1), max(x1), 1000)

# Интерполяция значений y
y_interp = np.interp(x_line, x1, y)

# Интерполяция погрешностей
t_interp = np.interp(x_line, x1, t)  

# Закрашивание области погрешностей
plt.fill_between(x_line, y_interp - t_interp, y_interp + t_interp, alpha=0.3, color='green', label='Погрешность')

# Построение точек без обводки
plt.scatter(x1, y, color='red', s=5, zorder=3)

# Линейная аппроксимация
a4, b = np.polyfit(x1, y, 1)


# Линия аппроксимации
y_line = a4 * x_line + b
plt.plot(x_line, y_line, label=f"y = {a4:.2f}x + {b:.2f}", color='blue')

# Добавление погрешностей к точкам без обводки
plt.errorbar(x1, y, yerr=t, fmt='o', color='red', capsize=0)

# Включение сетки
plt.grid()

# Подписи осей и заголовок
plt.xlabel("I, A")
plt.ylabel("B, Тл")
plt.title("Зависимость B от I (калибровочная кривая)")

plt.legend()

# Корректное сохранение файла
plt.savefig(r'C:\prog\Zavisimost_B_ot_I.pdf')

plt.show()
```
