```python
import numpy as np
import matplotlib.pyplot as plt

# Номера колец
n = np.arange(1, 20)

# Радиусы (уже с учётом нуля)
r_light = np.array([
    0.054, 0.108, 0.138, 0.164, 0.183, 0.202, 0.221, 0.239, 0.251,
    0.265, 0.279, 0.292, 0.304, 0.315, 0.327, 0.341, 0.351, 0.363, 0.372
])

r_dark = np.array([
    0.086, 0.125, 0.153, 0.174, 0.193, 0.212, 0.230, 0.244, 0.260,
    0.273, 0.287, 0.298, 0.310, 0.322, 0.336, 0.346, 0.357, 0.367, 0.377
])
# Квадраты радиусов
r2_light = r_light**2
r2_dark = r_dark**2

# Построение графика
plt.figure()

plt.scatter(n, r2_light, label='Светлые кольца')
plt.scatter(n, r2_dark, label='Тёмные кольца')

# Линейная аппроксимация
coef_light = np.polyfit(n, r2_light, 1)
coef_dark = np.polyfit(n, r2_dark, 1)

plt.plot(n, np.polyval(coef_light, n), linestyle='--')
plt.plot(n, np.polyval(coef_dark, n), linestyle='--')

plt.xlabel('Номер кольца n')
plt.ylabel(r'$r^2$')
plt.title('Зависимость $r^2$ от номера кольца')
plt.legend()
plt.grid()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Номера колец
n = np.arange(1, 20)

# Радиусы (уже /10)
r_light = np.array([
    0.054, 0.108, 0.138, 0.164, 0.183, 0.202, 0.221, 0.239, 0.251,
    0.265, 0.279, 0.292, 0.304, 0.315, 0.327, 0.341, 0.351, 0.363, 0.372
])

r_dark = np.array([
    0.086, 0.125, 0.153, 0.174, 0.193, 0.212, 0.230, 0.244, 0.260,
    0.273, 0.287, 0.298, 0.310, 0.322, 0.336, 0.346, 0.357, 0.367, 0.377
])

# Квадраты радиусов
r2_light = r_light**2
r2_dark = r_dark**2

# Линейная аппроксимация
coef_light = np.polyfit(n, r2_light, 1)
coef_dark = np.polyfit(n, r2_dark, 1)

k_light = coef_light[0]
k_dark = coef_dark[0]

plt.figure()

plt.scatter(n, r2_light, label='Светлые кольца')
plt.scatter(n, r2_dark, label='Тёмные кольца')

plt.plot(n, np.polyval(coef_light, n),
         linestyle='--',
         label=f'Аппрокс. светлых (k={k_light:.5f})')

plt.plot(n, np.polyval(coef_dark, n),
         linestyle='--',
         label=f'Аппрокс. тёмных (k={k_dark:.5f})')

plt.xlabel('Номер кольца n')
plt.ylabel(r'$r^2$, $мм^2$')
plt.title('Зависимость $r^2$ от номера кольца')
plt.legend()
plt.grid()
plt.savefig("r_2.jpg")

plt.show()
```
