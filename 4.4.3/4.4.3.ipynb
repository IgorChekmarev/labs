```python
import matplotlib.pyplot as plt
import numpy as np

x1 = np.array([690.7, 623.4, 579.1, 577, 546.1, 491.6, 435.8, 404.7])
y = np.array([1.6303, 1.6375, 1.6440, 1.6443, 1.6475, 1.6574, 1.6743, 1.6894])

t = 8*10**(-6)  # Погрешности

x_line = np.linspace(min(x1), max(x1), 1000)

# Интерполяция значений y
y_interp = np.interp(x_line, x1, y)

# Интерполяция погрешностей
#t_interp = np.interp(x_line, x1, t)

# Закрашивание области погрешностей относительно точек
#plt.fill_between(x_line, y_interp - t_interp, y_interp + t_interp, alpha=0.3, color='green', label='Погрешность')

# Построение точек
plt.scatter(x1, y, color='blue', s = 15, zorder=3)

#Построение апроксимации
a1, b = np.polyfit(x1, y, 1)
y_line = a1*x_line + b
plt.plot(x_line, y_line, label=f"y = {a1:.5f}x + {b:.3f}", color='green')

# Добавление погрешностей
#plt.errorbar(x1, y, xerr=t, yerr=t, linestyle='', color='black', capsize=2.5)

# Включение сетки
plt.grid()

plt.xlabel("$\lambda$, нм")
plt.ylabel("n")
plt.title("Зависимость n от $\lambda$")
plt.legend()
plt.savefig('Зависимость n от lambda.pdf')
plt.show()
```
